# Exploratory Data Analysis

### Import Libraries and Load Dataset

```python
import pandas as pd
import numpy as np
```

```python
from scipy.io import arff
```

```python
data = arff.loadarff("messidor_features.arff")
```

### Data Checks

```python
data[0]
```

```python
df = pd.DataFrame(data[0])
```

```python
df.head()
```

```python
df["Class"] = df["Class"].apply(lambda x: x.decode("utf-8") if isinstance(x, bytes) else x)
```

```python
df.head()
```

```python
df.describe()
```

```python
df.dtypes
```

```python
df["Class"] = pd.to_numeric(df["Class"]).astype(int)
```

```python
df.dtypes
```

```python
df.shape
```

```python
df.columns.tolist()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
duplicate_count = df.duplicated().sum()
duplicate_count
```

```python
df = df.drop_duplicates().reset_index(drop=True)
```

```python
df.shape
```

```python
# Export data to csv format for further processing
df.to_csv("messidor_features.csv", index=False)
```

### Target Class Distribution

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Calculate class counts
class_counts = df["Class"].value_counts().sort_index()
```

```python
# Plot class distribution
plt.figure(figsize=(7, 5))

ax = sns.countplot(
    data=df,
    x="Class",
    hue="Class",
    palette="Set2",
    legend=False
)

plt.title("Distribution of Target Classes")
plt.xlabel("Class")
plt.ylabel("Number of Observations")
plt.xticks([0, 1], ["Class 0", "Class 1"])

# Add count labels
for container in ax.containers:
    ax.bar_label(container)

plt.tight_layout()
plt.show()
```

### Input Features Distribution

```python
features = df.drop(columns=["Class"])
```

```python
features.hist(figsize=(16,14),
              bins=30,
              edgecolor="black")

plt.suptitle("Distribution of Features")
plt.tight_layout()
plt.show()
```

### Correlation Analysis

```python
corr_matrix = df.drop(columns=["Class"]).corr()
```

```python
plt.figure(figsize=(15,12))
sns.heatmap(corr_matrix,
            cmap="coolwarm",
            center=0,
            annot=True,
            square=True)

plt.title("Correlation Heatmap of Features")
plt.tight_layout()
plt.show()
```

### Outlier Analysis

```python
features=df.drop(columns=["Class"])
plt.figure(figsize=(16,9))
sns.boxplot(data=features, orient="h")
plt.xlabel("Feature Value")
plt.ylabel("Feature")
plt.tight_layout()
plt.show()
```
